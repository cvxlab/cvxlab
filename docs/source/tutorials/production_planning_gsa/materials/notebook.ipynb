# Tutorial - Production planning model
Complete workflow for definining and solving a simple production planning under 
resource constraints. Step by step guide aligned with User Guide. Ideal for newbies.

Package import

```python
import cvxlab
```

Generation of model directory

```python
cvxlab.create_model_dir(
    settings_file_type='yml', # or xlsx
    model_dir_name="test"
)
```

Fill model setup file(s) ...

Model class instance generation

```python
model = cvxlab.Model(
    log_level='debug',
    model_settings_from='yml', # or 'xlsx'
    use_existing_data=False,
    multiple_input_files=False,
    input_data_files_type="xlsx", # or "csv"
    detailed_validation=True,
    uncertainty=True
)
```

Fill sets data (model coordinates) ...

Initialization of data structures

```python
model.initialize_model_environment()
```

```python

model.uncertainty_settings(
    sampling_method="morris",
    gsa_method="morris",
    file_format="xlsx",
    sampling_kwargs={
        "N": 100,
        "num_levels": 4,
        "optimal_trajectories": 10,
        "seed": 123,
    },
)
```

```python
model.run_uncertainty(
       n_parallel=4,
   )
```
